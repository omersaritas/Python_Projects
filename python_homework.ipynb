```python
def flatten(lst):
    sonuc = []
    for i in lst:
        if isinstance(i, list):
            sonuc.extend(flatten(i))
        else:
            sonuc.append(i)
    return sonuc
print(flatten([[1, 'a', ['cat'], 2], [[[3]], 'dog'], 4, 5]))
```

```python
def reverse(lst):
    sonuc = []
    for eleman in reversed(lst):
        if isinstance(eleman, list):
            sonuc.append(reverse(eleman))
        else:
            sonuc.append(eleman)
    return sonuc


print(reverse([[1, 2], [3, 4], [5, 6, 7]]))
```
